## 3. Introduction to Machine Learning

### 3.1 What is Machine Learning?

Machine Learning merupakan pendekatan yang memungkinkan
komputer mempelajari pola dari data untuk menghasilkan prediksi
atau keputusan tanpa harus menentukan seluruh aturan secara
eksplisit.

Dalam Machine Learning, model mempelajari hubungan atau pola
yang terdapat pada data training dan kemudian menggunakan pola
tersebut untuk membuat prediksi terhadap data baru.

### 3.2 Traditional Programming vs Machine Learning

Pada pemrograman tradisional, programmer menentukan aturan
yang digunakan untuk menghasilkan output dari suatu input.

Pada Machine Learning, data dan hasil yang diharapkan digunakan
untuk melatih model sehingga model dapat mempelajari pola yang
terdapat pada data tersebut.

## 4. Supervised Learning

Supervised Learning merupakan jenis Machine Learning yang
menggunakan data yang memiliki target atau label.

Model mempelajari hubungan antara input dan target berdasarkan
data training. Setelah proses training selesai, model dapat
digunakan untuk memprediksi target dari data yang belum pernah
dilihat sebelumnya.

Dua permasalahan utama dalam supervised learning adalah:

1. Classification
2. Regression

### Classification

Classification digunakan ketika target yang ingin diprediksi
berupa kategori.

Contohnya adalah memprediksi apakah sebuah email termasuk
spam atau bukan spam.

### Regression

Regression digunakan ketika target yang ingin diprediksi
berupa nilai numerik.

Contohnya adalah memprediksi harga sebuah rumah.

## 5. Unsupervised Learning

Unsupervised Learning merupakan pendekatan Machine Learning
yang menggunakan data tanpa target atau label.

Tujuannya adalah menemukan pola atau struktur yang terdapat
di dalam data.

Salah satu contoh permasalahan unsupervised learning adalah
clustering, yaitu mengelompokkan data berdasarkan kemiripan
karakteristiknya.

## 7. The Iris Dataset

Dataset Iris merupakan dataset yang digunakan untuk
memperkenalkan proses pembangunan model Machine Learning.

Dataset ini berisi karakteristik bunga Iris yang dapat
digunakan untuk melakukan klasifikasi berdasarkan beberapa
fitur bunga.

```python
from sklearn.datasets import load_iris

iris_dataset = load_iris()
```

```python
print(iris_dataset.keys())
```

### Explanation

Objek `iris_dataset` berisi informasi yang berkaitan dengan
dataset Iris.

Beberapa informasi penting di dalamnya adalah:

- `data` berisi nilai fitur dari setiap sampel.
- `target` berisi label atau kelas setiap sampel.
- `target_names` berisi nama kelas.
- `feature_names` berisi nama fitur yang digunakan.
- `DESCR` berisi deskripsi dataset.

```python
print(iris_dataset['target_names'])
```

```python
print(iris_dataset['feature_names'])
```

```python
print(iris_dataset['data'].shape)
```

```python
print(iris_dataset['target'].shape)
```

### Features and Target

Feature merupakan variabel atau karakteristik yang digunakan
sebagai input untuk model Machine Learning.

Pada dataset Iris, fitur yang digunakan berkaitan dengan
karakteristik bunga.

Target merupakan nilai yang ingin diprediksi oleh model.
Pada kasus ini, target menunjukkan kelas atau jenis bunga Iris.

## 8. Training and Testing Data

Sebelum membangun model, dataset perlu dibagi menjadi training
data dan testing data.

Training data digunakan untuk melatih model agar mempelajari
pola dari data.

Testing data digunakan untuk mengevaluasi kemampuan model
terhadap data yang tidak digunakan selama proses training.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset['data'],
    iris_dataset['target'],
    random_state=0
)
```

### Explanation

Variabel `X` digunakan untuk menyimpan fitur atau input,
sedangkan `y` digunakan untuk menyimpan target atau label.

Setelah dilakukan train-test split:

- `X_train` berisi fitur untuk training.
- `X_test` berisi fitur untuk testing.
- `y_train` berisi target untuk training.
- `y_test` berisi target untuk testing.

Parameter `random_state=0` digunakan agar pembagian data dapat
direproduksi dan menghasilkan pembagian yang sama ketika kode
dijalankan kembali.

## 9. Building a K-Nearest Neighbors Model

K-Nearest Neighbors atau KNN merupakan algoritma supervised
learning yang melakukan prediksi berdasarkan data training
yang memiliki jarak paling dekat dengan data yang akan
diprediksi.

Parameter `n_neighbors` menentukan jumlah tetangga yang
digunakan oleh model.

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=1)
```

```python
knn.fit(X_train, y_train)
```

### Explanation

Method `fit()` digunakan untuk melatih model menggunakan
training data.

Pada kode tersebut:

- `X_train` merupakan fitur training.
- `y_train` merupakan target atau label training.

Setelah proses `fit()` selesai, model dapat digunakan untuk
melakukan prediksi terhadap data baru.

## 10. Making Predictions

```python
import numpy as np

X_new = np.array([[5, 2.9, 1, 0.2]])

prediction = knn.predict(X_new)

print("Prediction:", prediction)
print("Predicted target name:",
      iris_dataset['target_names'][prediction])
```

### Explanation

`X_new` merupakan data baru yang ingin diklasifikasikan.

Method `predict()` digunakan untuk menghasilkan prediksi
berdasarkan pola yang telah dipelajari model.

Hasil prediksi berupa label numerik yang kemudian dapat
dikonversi menjadi nama kelas menggunakan `target_names`.

## 11. Evaluating the Model

```python
test_score = knn.score(X_test, y_test)

print("Test set score:", test_score)
```

### Explanation

Method `score()` digunakan untuk mengukur performa model
terhadap testing data.

Pada classification, score yang digunakan oleh
`KNeighborsClassifier` adalah accuracy.

Accuracy menunjukkan proporsi data testing yang berhasil
diklasifikasikan dengan benar.

## 12. Results and Analysis

Model K-Nearest Neighbors menghasilkan test set score sebesar
0.9737 atau sekitar 97,37%.

Hasil tersebut menunjukkan bahwa sebagian besar data pada
testing set berhasil diklasifikasikan dengan benar.

Namun, nilai tersebut diperoleh dari satu pembagian training
dan testing data dengan `random_state=0`. Oleh karena itu,
hasil ini tidak dapat langsung dianggap sebagai performa model
pada seluruh kemungkinan data.
